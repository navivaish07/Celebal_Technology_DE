# Step 8: Spark Architecture

## Objective
Understand the architecture of Apache Spark and the roles of Driver, Cluster Manager, and Executors.

### Spark Architecture

Apache Spark follows a master-worker architecture.

- **Driver**
  - Creates the SparkSession.
  - Builds the DAG (Directed Acyclic Graph).
  - Schedules tasks.
  - Collects results.

- **Cluster Manager**
  - Allocates CPU and memory resources.
  - Launches Executors.

- **Executors**
  - Execute tasks.
  - Process partitions of data.
  - Return results to the Driver.

### Execution Modes

**Client Mode**
- Driver runs on the local machine.
- Best for development and testing.

**Cluster Mode**
- Driver runs inside the cluster.
- Best for production.

## Spark Architecture Diagram

             User Program
                  |
                  v
             Driver Program
                  |
                  v
           Cluster Manager
            /            \
           /              \
      Executor 1      Executor 2
          |                |
        Tasks            Tasks

## CSV vs Parquet

### CSV
- Row-based storage
- Larger file size
- Slower query performance
- No built-in compression
- Suitable for data exchange

### Parquet
- Columnar storage
- Smaller file size
- Faster analytical queries
- Supports compression
- Supports Predicate Pushdown

## Transformations

Transformations create a new DataFrame but do not execute immediately.

Examples:
- filter()
- select()
- withColumn()
- withColumnRenamed()

## Actions

Actions trigger the execution of all pending transformations.

Examples:
- show()
- count()
- collect()
- first()
